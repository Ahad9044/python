```python
# positional arguments
def positional_arguments(*nums):
    print(type(nums))
    for num in nums :
      print(num)

positional_arguments(1,2,3,4,5,6)
```

```python
# keywords arguments

def positional_arguments(**kwargs):
   
    for [key,value] in kwargs.items() :
      print(f"{key} : {value}")

positional_arguments(name:"ahad" , age: "23")
```

```python
# Series
# Series from list
import pandas as pd

a=[1,2,3,4,5]
series=pd.Series(a)
print(series)
```

```python
# Series from Dictionary
a= {1:"a" , 2:"b" , 3:"c", 4:"d"}
ser= pd.Series(a)
print(ser)
```

```python
index= [1,2,34,5]
data=['a', 'b' , 'c', 'd']
sr=pd.Series(index , index=data)
print(sr)
```
